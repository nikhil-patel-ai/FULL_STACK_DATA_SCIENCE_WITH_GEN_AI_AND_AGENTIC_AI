## Example

This starter example is self-contained and uses sample data where practical. It complements the original files in this folder; it does not replace them.


```python
from pathlib import Path

root = Path.cwd()
extensions = {".pdf", ".docx", ".pptx", ".xlsx", ".csv", ".ipynb", ".py"}
print("Reference files in the current folder and subfolders:")
for path in sorted(root.rglob("*")):
    if path.is_file() and path.suffix.lower() in extensions:
        print(f"{path.suffix.lower():7} {path.relative_to(root)}")
```
