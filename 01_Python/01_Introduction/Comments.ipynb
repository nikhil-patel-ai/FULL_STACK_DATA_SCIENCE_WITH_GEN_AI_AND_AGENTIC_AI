# Python Comments

A **comment** is a note in source code for people reading or maintaining the program. Python ignores comments when it runs the code. A comment starts with the hash symbol (`#`).

## 1. Single-Line Comments

Use `#` to write a comment. Python ignores everything from `#` to the end of that line.

```python
# Store the test score for later use.
score = 90

# Add a short explanation after a statement when it provides useful context.
passed = score >= 50  # A score of 50 or higher is a pass
print(passed)
```

## 2. Multi-Line Comments

Python has no special multi-line comment syntax. To comment out several lines, begin each line with `#`.

```python
# A multi-line comment uses one # on each line.
# This example adds the item price and tax.
item_price = 20
tax = 2
total = item_price + tax
print(total)
```

## 3. Docstrings and Triple-Quoted Strings

Triple-quoted text (`'''...'''` or `"""..."""`) is a string, not a comment. When a string is the first statement in a function, class, or module, it is called a **docstring** and documents that object. Use `#` for comments.

```python
def greet(name):
    """Return a friendly greeting for the given name."""
    return f"Hello, {name}!"


print(greet("Nikhil"))
```

## 4. Writing Helpful Comments

- Explain **why** a decision was made, not only what an obvious line of code does.
- Keep comments short, clear, and up to date with the code.
- Use comments to clarify intent or important details; prefer readable code over excessive comments.
