# Introduction to Python

## What Is Python?

Python is a high-level, general-purpose programming language known for its clear, readable syntax. It is approachable for beginners and powerful enough for professional software development. Python is often described as an interpreted language because a Python runtime executes Python programs.

Python is used in many fields, including:

- Data analysis, visualization, and scientific computing
- Artificial intelligence and machine learning
- Automation and scripting
- Web and application development

Python runs on Windows, macOS, and Linux, and its large community maintains a wide range of reusable libraries and learning resources.

## Set Up Your Python Tools

Python is the language runtime that executes your code. Anaconda is a Python distribution that includes Python and tools for managing environments and packages. VS Code is the editor where you will write and run code. In this lesson, we will use Anaconda with VS Code.

### 1. Install Anaconda

1. Download the installer for your operating system from the [official Anaconda download page](https://www.anaconda.com/download).
2. Run the installer and follow the setup prompts. The default options are suitable for getting started.
3. Open **Anaconda Prompt** on Windows, or a terminal on macOS/Linux, and verify the installation:

```text
python --version
conda --version
```

Each command should display a version number.

### 2. Install Visual Studio Code

1. Download the installer for your operating system from the [official VS Code download page](https://code.visualstudio.com/download).
2. Run the installer and follow the setup prompts.
3. In VS Code, open **Extensions**, search for **Python**, and install the extension published by Microsoft. If you plan to work with notebooks, install Microsoft's **Jupyter** extension as well.

### 3. Select Your Anaconda Environment

For a Python file, open the Command Palette and choose **Python: Select Interpreter**, then select your Anaconda environment. For a notebook, choose **Select Kernel** in the upper-right corner and select that same environment.
