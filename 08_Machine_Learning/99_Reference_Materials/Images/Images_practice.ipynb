## Example

This starter example is self-contained and uses sample data where practical. It complements the original files in this folder; it does not replace them.


```python
from pathlib import Path

folder = Path.cwd()
print("Current folder:", folder)
files = sorted(p for p in folder.iterdir() if p.is_file())
folders = sorted(p for p in folder.iterdir() if p.is_dir())

print("\nFiles:")
for item in files:
    print(f"- {item.name} ({item.stat().st_size:,} bytes)")
print("\nSubfolders:")
for item in folders:
    print(f"- {item.name}")

# Open data or documents only after checking their file type and contents.
```
