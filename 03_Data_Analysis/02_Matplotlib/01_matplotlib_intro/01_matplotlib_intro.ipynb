## Topic outline
- **What is Matplotlib?:** Matplotlib is a Python library for creating charts and visualizations.
- **Where is the Matplotlib Codebase?:** The source code is maintained in the Matplotlib GitHub repository.


```python
import matplotlib
import matplotlib.pyplot as plt

print("Matplotlib version:", matplotlib.__version__)

x = [1, 2, 3, 4]
y = [10, 15, 13, 20]

plt.plot(x, y)
plt.title("My First Matplotlib Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```
