## Topic outline
- **What is Pyplot?:** Pyplot is Matplotlib's state-based interface for making charts with simple function calls.
- **Plotting x and y points:** Pass x-coordinates and y-coordinates to `plt.plot()` to draw data.
- **The x-axis:** The x-axis usually represents the independent variable or categories.
- **The y-axis:** The y-axis usually represents the measured values.
- **Plotting Without Line:** Use a format string such as `o` to show points without connecting lines.


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [5, 7, 4, 8]

plt.plot(x, y, "o")
plt.title("Pyplot Points")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
