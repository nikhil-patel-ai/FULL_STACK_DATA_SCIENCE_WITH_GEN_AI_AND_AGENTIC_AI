# Bars

Use bar charts to compare values across discrete categories.

Source: https://www.w3schools.com/python/matplotlib_bars.asp

## Learning objective
Understand the concept, run the example, then change the data and try again.

## Topic outline
- **Creating Bars:** `plt.bar()` creates vertical bars for category comparisons.
- **Horizontal Bars:** `plt.barh()` creates horizontal bars, useful for long category labels.
- **Bar Color:** The `color` argument sets the color of bars.
- **Bar Width:** The `width` argument controls the width of vertical bars.


```python
import matplotlib.pyplot as plt

subjects = ["Python", "Math", "Statistics", "ML"]
hours = [8, 5, 6, 4]

plt.bar(subjects, hours, color="teal", width=0.6)
plt.title("Weekly Study Hours")
plt.xlabel("Subject")
plt.ylabel("Hours")
plt.show()

plt.figure()
plt.barh(subjects, hours, color="coral")
plt.title("Weekly Study Hours (Horizontal)")
plt.xlabel("Hours")
plt.ylabel("Subject")
plt.show()
```
