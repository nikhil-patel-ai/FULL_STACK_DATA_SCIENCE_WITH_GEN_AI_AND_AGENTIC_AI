# Histograms

Use histograms to understand how numerical observations are distributed across intervals.

Source: https://www.w3schools.com/python/matplotlib_histograms.asp

## Learning objective
Understand the concept, run the example, then change the data and try again.

## Topic outline
- **Histogram:** A histogram groups numeric observations into bins and displays their frequencies.
- **Create Histogram:** Use `plt.hist()` to divide values into bins and count observations in each bin.


```python
import matplotlib.pyplot as plt

exam_scores = [42, 55, 58, 61, 62, 65, 68, 70, 72, 72, 75, 78, 81, 85, 90, 92]

plt.hist(exam_scores, bins=5, edgecolor="black")
plt.title("Distribution of Exam Scores")
plt.xlabel("Score range")
plt.ylabel("Number of students")
plt.show()
```
