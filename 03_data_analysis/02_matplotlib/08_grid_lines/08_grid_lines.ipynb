# Grid Lines

Add grid lines to help readers estimate values from a chart.

Source: https://www.w3schools.com/python/matplotlib_grid.asp

## Learning objective
Understand the concept, run the example, then change the data and try again.

## Topic outline
- **Add Grid Lines to a Plot:** Call `plt.grid()` to display grid lines on the chart.
- **Specify Which Grid Lines to Display:** Choose `axis='x'`, `axis='y'`, or `axis='both'` to control the grid direction.
- **Set Line Properties for the Grid:** Style grid lines with options such as color, linestyle, and linewidth.


```python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [12, 17, 14, 21, 25]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(axis="y", color="gray", linestyle="--", linewidth=0.6)
plt.show()
```
