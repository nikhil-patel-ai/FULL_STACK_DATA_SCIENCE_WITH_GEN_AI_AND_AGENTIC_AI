# Pie Charts

Use pie charts to show how categories contribute to a whole.

Source: https://www.w3schools.com/python/matplotlib_pie_charts.asp

## Learning objective
Understand the concept, run the example, then change the data and try again.

## Topic outline
- **Creating Pie Charts:** `plt.pie()` displays values as proportional slices of a circle.
- **Labels:** The `labels` argument names each slice.
- **Start Angle:** The `startangle` argument rotates the chart to choose where the first slice begins.
- **Explode:** The `explode` argument offsets selected slices from the center.
- **Shadow:** Set `shadow=True` to add a shadow behind the pie chart.
- **Colors:** The `colors` argument assigns colors to slices.
- **Legend:** A legend identifies the categories represented by the slices.
- **Legend With Header:** Use `plt.legend(title=...)` to add a heading to the legend.


```python
import matplotlib.pyplot as plt

labels = ["Python", "Statistics", "Math", "ML"]
hours = [8, 5, 4, 6]
explode = [0.08, 0, 0, 0]

plt.pie(
    hours,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow=True
)
plt.title("Study Time by Subject")
plt.legend(title="Subjects", loc="best")
plt.show()
```
