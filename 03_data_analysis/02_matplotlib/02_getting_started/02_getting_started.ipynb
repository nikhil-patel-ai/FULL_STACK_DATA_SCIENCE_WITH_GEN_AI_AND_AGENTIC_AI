## Topic outline
- **Installation of Matplotlib:** Install the library with pip before importing it.
- **Import Matplotlib:** Import pyplot using the common alias `plt` to create charts.
- **Checking Matplotlib Version:** Print `matplotlib.__version__` to see which version is installed.


```python
import matplotlib
import matplotlib.pyplot as plt

print("Matplotlib version:", matplotlib.__version__)

plt.plot([1, 2, 3], [2, 4, 6])
plt.title("Installation Check")
plt.show()
```
