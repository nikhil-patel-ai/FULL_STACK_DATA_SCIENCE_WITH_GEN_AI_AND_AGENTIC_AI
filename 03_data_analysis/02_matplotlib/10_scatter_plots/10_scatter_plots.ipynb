# Scatter Plots

Use scatter plots to examine relationships between two numeric variables.

Source: https://www.w3schools.com/python/matplotlib_scatter.asp

## Learning objective
Understand the concept, run the example, then change the data and try again.

## Topic outline
- **Creating Scatter Plots:** `plt.scatter()` draws each x-y pair as an individual point.
- **Compare Plots:** Plot multiple datasets with different colors to compare groups.
- **Colors:** The `color` or `c` argument controls point colors.
- **Color Each Dot:** Provide a sequence of colors or numeric values to vary point colors.
- **ColorMap:** A colormap maps numeric values to a range of colors.
- **Size:** The `s` argument controls marker area.
- **Alpha:** The `alpha` argument controls transparency and helps reveal overlapping points.


```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
scores = [45, 50, 58, 65, 73, 88]
sizes = [40, 70, 90, 110, 140, 180]

plt.scatter(study_hours, scores, s=sizes, c=scores, cmap="viridis", alpha=0.8)
plt.colorbar(label="Score")
plt.title("Study Hours and Scores")
plt.xlabel("Study hours")
plt.ylabel("Score")
plt.show()
```
