# Generate a numeric one-time password (OTP)

This example uses Python's `secrets` module for secure random choices. The OTP is returned as a string so leading zeroes are preserved. This is a generator example, not a complete authentication system.

```python
import secrets
import string


def generate_otp(length=6):
    """Generate a cryptographically secure numeric OTP."""
    if isinstance(length, bool) or not isinstance(length, int) or length <= 0:
        raise ValueError("length must be a positive integer")
    return "".join(secrets.choice(string.digits) for _ in range(length))


otp = generate_otp()
print(f"Your OTP is: {otp}")
```
